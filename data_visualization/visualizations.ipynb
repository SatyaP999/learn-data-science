```python
from matplotlib import pyplot as plt
import pandas as pd
from collections import defaultdict
```

```python
col_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
```

```python
iris_data = pd.read_csv('iris.data', header=None, names=col_names)
```

```python
print(iris_data.head(5))
```

```python
print(iris_data.tail(5))
```

```python
species = iris_data['species']
print(species)
```

```python
species_count = defaultdict(int)
for s in species:
    species_count[s] += 1

species_count
    
```

```python
count_lst = []
species_lst = []
for key in species_count.keys():
    species_lst.append(key)
    count_lst.append(species_count[key])

count_lst, species_lst
```

```python
plt.bar(range(len(species_lst)), count_lst)

plt.title("Iris species distribution")
plt.ylabel("Count of each species")

plt.xticks((range(len(species_lst))), species_lst)

plt.show()
```

```python
sepal_length = iris_data['sepal_length']
petal_length = iris_data['petal_length']


plt.scatter(sepal_length, petal_length)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")

plt.axis("equal")

plt.show()
```

```python
plt.hist(sepal_length, bins=15, color='skyblue', edgecolor='black') # Plot histogram
plt.title('Histogram of Iris Sepal Length', fontsize=14) # Add title
plt.xlabel('Sepal Length (cm)', fontsize=12) # Add x-axis label
plt.ylabel('Frequency', fontsize=12) # Add y-axis label
plt.show()
```
