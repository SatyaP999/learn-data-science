```python
def vector_add(vec_1: list, vec_2: list) -> list | str:
    """Adds two vectors of equal length"""
    if len(vec_1) == len(vec_2):
        result = [vec_1_i + vec_2_i for vec_1_i, vec_2_i in zip(vec_1, vec_2)]
        return result
    else:
        return "Both vectors should be of equal length."
```

```python
print(vector_add([1, 2], [3, 4]))
```

```python
print(vector_add([12, 3, 4], [4, 5]))
```

```python
def vector_sub(vec_1: list, vec_2: list) -> list | str:
    """Subtracts two vectors of equal length"""
    if len(vec_1) == len(vec_2):
        result = [vec_1_i - vec_2_i for vec_1_i, vec_2_i in zip(vec_1, vec_2)]
        return result
    else:
        return "Both vectors should be of equal length."
```

```python
print(vector_sub([1, 2], [3, 4]))
```

```python
print(vector_sub([12, 3, 4], [4, 5]))
```

```python
def vector_dot(vec_1: list, vec_2: list) -> float:
    """Returns dot product of two vectors"""
    if len(vec_1) == len(vec_2):
        result = sum([vec_1_i * vec_2_i for vec_1_i, vec_2_i in zip(vec_1, vec_2)])
        return result
    else:
        return "Both vectors should be of equal length."
```

```python
print(vector_dot([1, 2], [3, 4]))
```

```python
print(vector_dot([12, 3, 4], [4, 5]))
```
